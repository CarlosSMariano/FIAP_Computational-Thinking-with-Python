Operadores Aritméticos:


São responsáveis por realizar operações matemáticas, são eles:

    Soma (+);
    Subtração (–);
    Divisão (/) "Retorna Double";
    Multiplicação (*);
    Módulo (%) "Retorna o resto de uma divisão";
    Parte inteira da divisão (//);

```python

num1 = 10
num2 = 2

soma = num1 + num2
sub = num1 - num2
divisao = num1 / num2
mult = num1 * num2
mod = num1 % num2
parte_inteira = num1 // num2

print(f"""
    Soma: {num1} + {num2} = {soma}
    Subtração: {num1} – {num2} = {sub}
    Divisão: {num1} / {num2} = {divisao}
    Multiplicação: {num1} * {num2} = {mult}
    Módulo: {num1} % {num2} = {mod}
    Parte inteira da divisão: {num1} // {num2} = {parte_inteira}
""")
```

Operadores de atribuição (assignment), incremento e decremento:

Os operadores responsáveis por atribuir valores à váriaveis são:

Comumente utilizados:

    "=" Atribui um valor para a variável
    "+=" Soma a variável com outro valor

Menos utilizados:

    "-=" Subtra a variável com outro valor
    "*=" Multiplica a variável com outro valor
    "/=" Divide a variável com outro valor

Não é utilizada em outras linguagens (ignore):
    
    ":=" Atribui um valor para a variável dentro de uma expressão

Python não utlizia operadores de incremento e decremento convencionais "++" e "--", pra isso utilizamos os de atribuição "+=" e "-="

```python

i = 0
i += 10
print(i)
i -= 2
print(i)
i *= 4
print(i)
i /= 2
print(i)
```

Operadores de comparação:

São utilizados para comparar dois valores e retornam um resultado booleano (True ou False), são eles:

    == igual;
    != diferente;
    >  maior que;
    <  menor que;
    >= maior ou igual;
    <= menor ou igual;


```python

a = 2
b = 3
print(f"A comparação {a} > {b} resulta em {a>b}")
print(f"A comparação {a} >= {b} resulta em {a>=b}")
print(f"A comparação {a} < {b} resulta em {a<b}")
print(f"A comparação {a} <= {b} resulta em {a<=b}")
print(f"A comparação {a} == {b} resulta em {a==b}")
print(f"A comparação {a} != {b} resulta em {a!=b}")
```

Operadores lógicos:

Os operadores lógicos são utilizados para comparar multiplas condições em uma expressão. São eles:

    and: Retorna True se todas as condições forem verdadeiras. Caso qualquer uma das condições seja falsa, ele retorna False;

    or: Retorna True se qualquer uma das condições for verdadeira. Ele só retorna False se todas as condições forem falsas;

    not: Inverte o valor de uma condição. Se a condição for True, ele retorna False. Ele é equivalente ao sinal;

```python

a = 2
b = 3
print("TABELINHA DO AND:")
print(f"{a} > {b} and {a} < {b}, ou seja, {a>b} and {a<b} dá {a>b and a<b}")
print(f"{a} == {b} and {a} > {b}, ou seja, {a==b} and {a>b} dá {a==b and a>b}")
print(f"{a} < {b} and {a} > {b}, ou seja, {a<b} and {a>b} dá {a<b and a>b}")
print(f"{a} != {b} and {a} < {b}, ou seja, {a!=b} and {a<b} dá {a!=b and a<b}")
print()
print("TABELINHA DO OR:")
print(f"{a} > {b} or {a} < {b}, ou seja, {a>b} or {a<b} dá {a>b or a<b}")
print(f"{a} == {b} or {a} > {b}, ou seja, {a==b} or {a>b} dá {a==b or a>b}")
print(f"{a} < {b} or {a} > {b}, ou seja, {a<b} or {a>b} dá {a<b or a>b}")
print(f"{a} != {b} or {a} < {b}, ou seja, {a!=b} or {a<b} dá {a!=b or a<b}")
```

Operador de Identidade:

O operador de identidade "is" verifica se dois objetos (listas etc) são o mesmo objeto na memória (ou seja, ocupam o mesmo endereço de memória), em vez de apenas comparar seus valores.

```python

a = [1, 2, 3]
b = a
c = [1, 2, 3]
print("a e b são o mesmo objeto:")
print(a is b)
print("a e c são o mesmo objeto:")
print(a is c)
```

Operador de Pertencimento (Membership):

O operador "in" é usado para verificar se um elemento está presente em uma coleção, como listas, dicionários ou strings. Esse operador é bastante intuitivo e amplamente utilizado, especialmente para validação de dados.

```python

frutas = ["Abacate", "Tomate", "Carambola", "Acerola", "Graviola"]

print("Tomate é uma fruta:")
print("Tomate" in frutas)

print("Cebola é uma fruta:")
print("Cebola" in frutas)
```
