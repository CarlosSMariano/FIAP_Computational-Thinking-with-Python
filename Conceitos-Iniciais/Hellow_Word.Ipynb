A função "type(objeto)" retorna o tipo de dado do objeto representado como uma classe

Python é uma linguagem com Tipagem Dinâmica (linguagem dinâmica), ou seja, não precisamos declarar o tipo do dado que a váriavel irá armazenar, a própria linguagem inferi o tipo a partir do valor atribuido.

Todavia, uma vez que uma variável pertence a um tipo, não podemos alterar seu valor de forma implícita, nem realizar uma operação que não é suportada pelo tipo de variável. Python lançará uma exeção, a menos que você converta o valor explicitamente.

```python

palavra_1 = "Hello "
print(type(palavra_1))

palavra_2 = 12
print(type(palavra_2))

frase = palavra_1 + '+' + palavra_2
print(type(frase))
```

INPUT:

A função "input()" permite que um programa receba dados diretamente do usuário durante sua execução.
Já a função "int()" converte o objeto em inteiro, caso não consiga executar a operação o python retorna uma exeção. 

```python

palavra_1 = input("Diga a primeira palavra: ")
palavra_2 = input("Diga a segunda palavra: ")
palavra_3 = input("Diga a terceira palavra: ")
palavra_4 = input("Diga a quarta palavra: ")
frase = f"""
palavra1: {palavra_1} 
palavra1: {palavra_2} 
palavra1: {palavra_3} 
palavra1: {palavra_4}
"""
print(frase)

numero1 = int(input("Digite um número"))
print(numero1)
```
